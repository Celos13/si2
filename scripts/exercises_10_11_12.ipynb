```python
import numpy as np
from si.neural_networks.layers import Dropout

np.random.seed(0)

X = np.random.randn(4, 6)
drop = Dropout(probability=0.5)

out_train = drop.forward_propagation(X, training=True)
out_test = drop.forward_propagation(X, training=False)

print("Zeros no treino:", (out_train == 0).sum())
print("Inferência igual ao input:", np.allclose(out_test, X))
```
