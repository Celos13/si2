# Exercício 1

```python
from pathlib import Path
import numpy as np

from si.io.csv_file import read_csv
from si.data.dataset import Dataset
from si.feature_selection.select_percentile import SelectPercentile
from si.statistics.f_classification import f_classification

iris_path = Path(r"C:\Users\filip\OneDrive\Attachments\Ambiente de Trabalho\SIB_gh\si2\datasets\iris\iris.csv")
iris_ds = read_csv(iris_path, sep=",", features=True, label=True)

X, y = iris_ds.X, iris_ds.y
print("Shape X:", X.shape)
print("Classes:", np.unique(y, return_counts=True))
print("Features:", iris_ds.features)
print("Label name:", iris_ds.label)
```

**Exercício 1.3 – Últimas 10 amostras e média por feature**

```python
last10 = X[-10:, :]
mean_last10 = np.mean(last10, axis=0)

print("Últimas 10 amostras:\n", last10)
print("Média das últimas 10 amostras por feature:", mean_last10)
```

**Exercício 1.4 – Amostras com todas as features <= 6**

```python
mask_less6 = np.all(X <= 6, axis=1)
num_less6 = np.sum(mask_less6)

print("Número de amostras com todas as features <= 6:", num_less6)
```

**Exercício 1.5 – Amostras com label != 'Iris-setosa'**

# Exercício 2

```python
from si.data.dataset import Dataset
import numpy as np

X_demo = np.array([
    [1.0, 2.0],
    [np.nan, 3.0],
    [4.0, np.nan],
    [5.0, 6.0]
])
y_demo = np.array([0, 1, 1, 0])

demo_ds = Dataset(X=X_demo.copy(), y=y_demo.copy(), features=["f1", "f2"], label="class")
print("Dataset de exemplo (com NaNs):\n", demo_ds.X)
```

**Exercício 2 – fillna com média**

```python
demo_fill = Dataset(
    X=X_demo.copy(),
    y=y_demo.copy(),
    features=["f1", "f2"],
    label="class"
)

demo_fill.fillna("mean")
print("Depois de fillna('mean'):\n", demo_fill.X)
```

**Exercício 2 – dropna**

```python
demo_drop = Dataset(
    X=X_demo.copy(),
    y=y_demo.copy(),
    features=["f1", "f2"],
    label="class"
)

demo_drop.dropna()
print("Depois de dropna():")
print("X:\n", demo_drop.X)
print("y:", demo_drop.y)
```

**Exercício 2 – remove_by_index**

```python
demo_rm = Dataset(
    X=X_demo.copy(),
    y=y_demo.copy(),
    features=["f1", "f2"],
    label="class"
)

demo_rm.remove_by_index([1, 2])
print("Depois de remover índices [1, 2]:")
print("X:\n", demo_rm.X)
print("y:", demo_rm.y)
```

# Exercício 3

**Exercício 3 – SelectPercentile com f_classification no iris**

```python
selector = SelectPercentile(score_func=f_classification, percentile=50)  
selector.fit(iris_ds)
iris_reduced = selector.transform(iris_ds)

print("Shape original:", iris_ds.X.shape)
print("Shape reduzido:", iris_reduced.X.shape)
print("Features selecionadas:", iris_reduced.features)
```
