## Exercício 10 

```python

import os
from datasets import DATASETS_PATH
from si.io.data_file import read_data_file
from si.model_selection.split import train_test_split
from si.ensemble.stacking_classifier import StackingClassifier
from si.models.knn_classifier import KNNClassifier
from si.models.logistic_regression import LogisticRegression
from si.models.decision_tree_classifier import DecisionTreeClassifier

csv_file = os.path.join(DATASETS_PATH, "breast_bin", "breast-bin.csv")
ds = read_data_file(csv_file, label=True, sep=",")

train_ds, test_ds = train_test_split(ds, test_size=0.2, random_state=42)

m1 = KNNClassifier(k=3)
m2 = LogisticRegression(max_iter=2000)
m3 = DecisionTreeClassifier(max_depth=5)

final_model = KNNClassifier(k=3)

model = StackingClassifier(models=[m1, m2, m3], final_model=final_model)
model.fit(train_ds)

print("Train score:", model.score(train_ds))
print("Test score:", model.score(test_ds))
```

## Exercício 11 

```python

import numpy as np
from si.model_selection.randomized_search import randomized_search_cv
from si.models.logistic_regression import LogisticRegression

model = LogisticRegression()

param_dist = {
    "l2_penalty": np.linspace(1, 10, 10),
    "alpha": np.linspace(0.001, 0.0001, 10),
    "max_iter": np.linspace(500, 1500, 5),
}

results = randomized_search_cv(
    model=model,
    dataset=train_ds,
    hyperparameter_grid=param_dist,
    cv=3,
    n_iter=5,
    random_state=42
)

results
```
