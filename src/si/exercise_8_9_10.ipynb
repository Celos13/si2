```python
from pathlib import Path
import numpy as np

from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split  
from si.models.random_forest_classifier import RandomForestClassifier

iris_path = Path(r"C:\Users\filip\OneDrive\Attachments\Ambiente de Trabalho\SIB_gh\si2\datasets\iris\iris.csv")
dataset = read_csv(iris_path, sep=",", features=True, label=True)

print("Shape total:", dataset.X.shape)
print("Classes:", np.unique(dataset.y, return_counts=True))
```

```python
train_ds, test_ds = train_test_split(dataset, test_size=0.3, random_state=42)
```

```python
model = RandomForestClassifier(
    n_estimators=50,
    max_features=2,
    min_sample_split=2,
    max_depth=5,   # ou 10, por exemplo
    mode="gini",
    seed=42
)


model.fit(train_ds)

train_score = model.score(train_ds)
test_score = model.score(test_ds)

print("Accuracy train:", train_score)
print("Accuracy test :", test_score)
```

```python
from pathlib import Path

from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split
from si.models.knn_classifier import KNNClassifier
from si.models.logistic_regression import LogisticRegression
from si.models.decision_tree_classifier import DecisionTreeClassifier
from si.ensemble.stacking_classifier import StackingClassifier

breast_path = Path(r"C:\Users\filip\OneDrive\Attachments\Ambiente de Trabalho\SIB_gh\si2\datasets\breast_bin\breast-bin.csv")
dataset = read_csv(breast_path, sep=",", features=True, label=True)

train_ds, test_ds = train_test_split(dataset, test_size=0.3, random_state=42)

train_ds.X.shape, test_ds.X.shape
```

```python
knn1 = KNNClassifier(k=5)
logreg = LogisticRegression(l2_penalty=1, alpha=0.001, max_iter=1000)
tree = DecisionTreeClassifier(max_depth=5)
```

```python
final_knn = KNNClassifier(k=3)
```

```python
stack_model = StackingClassifier(
    models=[knn1, logreg, tree],
    final_model=final_knn
)
```

```python
stack_model.fit(train_ds)

train_score = stack_model.score(train_ds)
test_score = stack_model.score(test_ds)

print("Train Accuracy:", train_score)
print("Test Accuracy:", test_score)
```
